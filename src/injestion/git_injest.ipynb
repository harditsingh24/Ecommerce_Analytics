```python
import pandas as pd
from pyspark.sql import SparkSession
```

```python
# Read CSV into Pandas DataFrame
df = pd.read_csv(
    "https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/customers.csv"
)

# Create Spark Session
spark = SparkSession.builder.getOrCreate()

# Convert Pandas DataFrame to Spark DataFrame
spark_df = spark.createDataFrame(df)

# (Optional) Another Spark DataFrame reference
df_pandas = spark.createDataFrame(df)

# Save Spark DataFrame as CSV
spark_df.write \
    .format("csv") \
    .mode("overwrite") \
    .save("/Volumes/ecommerce_etl_updated/bronze/raw_data/customers")
```
